# Linear Regression

### 1. Import thư viện cần thiết và load dataset

```python
import os
import joblib
import pandas as pd
from sklearn.linear_model import LinearRegression

data_train = pd.read_csv("../data/preprocessed/data_train.csv")
data_test = pd.read_csv("../data/preprocessed/data_test.csv")

display(data_train)
display(data_test)
```

### 2. Định nghĩa các biến `FEATURES` và biến `TARGET`

```python
FEATURES = [
    "distance_m",
    "signal_mean_pct",
    "rx_link_rate_mean",
    "rtt_mean_ms",
    "jitter_ms",
    "packet_loss_pct"
]

TARGET = "throughput_down_mbps"

X_train = data_train[FEATURES]
y_train = data_train[TARGET]

X_test = data_test[FEATURES]
y_test = data_test[TARGET]
```

### 3. Train mô hình

```python
model = LinearRegression()
model.fit(X_train, y_train)

y_intercept = model.intercept_
coeff_table = pd.DataFrame({
    'Feature': model.feature_names_in_,
    'Coefficient': model.coef_
})

print("The y-intercept of the model is:", y_intercept)
display(coeff_table)
```

#### Ý nghĩa
- `distance_m`: Cứ mỗi 1 mét đi xa khỏi Access Point, thông lượng sẽ mất đi ~0.022 Mbps
- `signal_mean_pct`: Cường độ tín hiệu tăng thêm 1% thì thông lượng sẽ tăng thêm ~1.093 Mbps
- `rx_link_rate`: Tốc độ liên kết tăng thêm 1 Mbps thì thông lượng sẽ tăng thêm ~0.04 Mbps
- `rtt_mean_ms`: Thời gian trễ trọn vòng tăng thêm 1 mili-giây thì thông lượng sẽ giảm đi ~0.005 Mbps
- `jitter_ms`: Biến thiên độ trễ tăng thêm 1 mili-giây thì thông lượng sẽ giảm đi ~0.001 Mbps
- `packet_loss_pct`: Tỷ lệ mất gói tin tăng thêm 1% thì thông lượng sẽ giảm đi ~0.105 Mbps

### 4. Sử dụng mô hình đã train để dự đoán biến `TARGET` của tập `data_test`

```python
y_pred_linear_regression = model.predict(X_test)

y_pred_linear_regression = pd.DataFrame(
    y_pred_linear_regression,
    columns=["prediction_label"],
    index=X_test.index
)

display(y_pred_linear_regression)
```

### 5. Lưu lại dự đoán của mô hình Linear Regression

```python
try:
    y_pred_linear_regression.to_csv("../data/prediction-linear-regression/prediction_linear_regression.csv", mode="x", index=False)
    print("File created successfully.")
except FileExistsError:
    print("File already exists.")
```

### 6. Lưu lại mô hình

```python
model_path = "../models/linear_regression_model.joblib"

if not (os.path.exists(model_path)):
    joblib.dump(model, model_path)
    print("Model created successfully.")
else:
    print("File already exists.")
```
