# Baseline

### 1. Import thư viện cần thiết và load dataset

```python
import pandas as pd

data_train = pd.read_csv("../data/preprocessed/data_train.csv")
data_test = pd.read_csv("../data/preprocessed/data_test.csv")

display(data_train)
display(data_test)
```

### 2. Định nghĩa các biến `FEATURES` và biến `TARGET`

```python
FEATURES = [
    "distance_m",
    "signal_mean_pct",
    "rx_link_rate_mean",
    "rtt_mean_ms",
    "jitter_ms",
    "packet_loss_pct"
]

TARGET = "throughput_down_mbps"
```

### 3. Tính trung bình biến `TARGET` của tập `data_train`

```python
baseline_mean = data_train[TARGET].mean()

print(f"Training mean throughput: {baseline_mean:.2f} Mbps")
```

### 4. Sử dụng trung bình trên để dự đoán giá trị biến `TARGET` của tập `data_test`

```python
y_pred_baseline = data_test[[TARGET]]

y_pred_baseline[:] = baseline_mean

display(y_pred_baseline)
```

### 5. Lưu lại dự đoán của mô hình baseline

```python
try:
    y_pred_baseline.to_csv("../data/prediction-baseline/prediction_baseline.csv", mode="x", index=False)
    print("File created successfully.")
except FileExistsError:
    print("File already exists.")
```
