# Data Cleaning

### 1. Import thư viện pandas và load dataset

```python
import pandas as pd

data = pd.read_csv("../data/raw/data-raw.csv")

display(data)
```

### 2. Bỏ các cột `observation_id`, `session_id` và `time_of_day`

Lý do: Không sử dụng `observation_id` và `session_id` trong EDA hay các bước học máy sau. Trong khi đó `time_of_day` lại tính toán từ `timestamp` (Chương 3 báo cáo)

```python
data = data.drop(columns=["observation_id", "session_id", "time_of_day"])

data.head()
```

### 3. Đổi kiểu giá trị biến `timestamp` thành kiểu số, đồng thời đổi tên thành `seconds_since_midnight`
Lý do: Giúp dễ tính toán và khám phá trong EDA

```python
ts_col = pd.to_datetime(data['timestamp'])

data['timestamp'] = ts_col.dt.hour * 3600 + ts_col.dt.minute * 60 + ts_col.dt.second

data.rename(columns={'timestamp': 'seconds_since_midnight'}, inplace=True)

data.head()
```

### 4. Thay giá trị tương ứng các địa điểm thu dữ liệu thành giá trị khoảng cách, đồng thời đổi tên thành `distance_m`

```python
data['location_id'] = data['location_id'].replace({1: 1, 2: 15, 3: 5})
data.rename(columns={'location_id': 'distance_m'}, inplace=True)

data.head()
```

### 5. Đổi tên cột `signal_mean` và `signal_std` thành lần lượt `signal_mean_pct` và `signal_std_pct`

```python
data.rename(columns={'signal_mean': 'signal_mean_pct', 'signal_std': 'signal_std_pct'}, inplace=True)

data.head()
```

### 6. Lưu lại dataset đã qua cleaning

```python
try:
    data.to_csv("../data/cleaned/data-cleaned.csv", mode="x", index=False)
    print("File created successfully.")
except FileExistsError:
    print("File already exists.")
```
